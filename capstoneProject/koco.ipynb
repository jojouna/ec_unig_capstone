### Google Translation API

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv("GOOGLE_API_KEY")
print("Key loaded:", API_KEY is not None)
```

```python
URL = "https://translation.googleapis.com/language/translate/v2"

def translate_google(text, target_language="en", source_language="ko"):
    """
    Function for translating text using Google Translation API
    """

    # If the text is empty, return the text
    if pd.isna(text) or str(text).strip() == "":
        return text

    # Set the parameters
    params = {
        "q": text,
        "target": target_language,
        "source": source_language,
        "key": API_KEY
    }
    response = requests.post(URL, data=params)

    if response.status_code != 200:
        print(f"Error translating: {text[:50]}... -> {response.text}")
        return None

    result = response.json()
    return result["data"]["translations"][0]["translatedText"]

df_train_10['comments_trans_g_en'] = df_train_10['comments'].apply(lambda x: translate_google(x))
df_train_10['title_trans_g_en'] = df_train_10['news_title'].apply(lambda x: translate_google(x))
```

```python
pd.set_option('display.max_colwidth', None)  # no limit

df_train_10
```

```python
import html
cols_to_clean = ["comments_trans_g_en", "title_trans_g_en"]

for c in cols_to_clean:
    df_train_10[c] = df_train_10[c].apply(lambda x: html.unescape(x) if pd.notna(x) else x)
```
